# Encoding dos Rótulos de Sentimento

Este notebook apresenta a etapa de encoding dos rótulos de sentimento
das avaliações do IMDb.

## Objetivo

Transformar os rótulos textuais em valores numéricos para utilização
nas etapas posteriores de aprendizado de máquina.

### Mapeamento utilizado

- `negative` → `0`
- `positive` → `1`

Nesta etapa, apenas os rótulos de sentimento são codificados.
A representação numérica dos textos será realizada posteriormente.

```python
import sys
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT / "src") not in sys.path:
    sys.path.append(str(PROJECT_ROOT / "src"))

from preprocessing.encoding import encode_sentiment
```

```python
INPUT_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "IMDB_Dataset_normalizado.csv"
)

OUTPUT_FILE = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "IMDB_Dataset_encoded.csv"
)

print("Entrada:", INPUT_FILE)
print("Saída:", OUTPUT_FILE)
```

```python
df = pd.read_csv(INPUT_FILE)

print("Quantidade de linhas:", len(df))
print("Quantidade de colunas:", len(df.columns))

df.head()
```

```python
df["sentiment"].value_counts()
```

```python
df_encoded = df.copy()

df_encoded["sentiment"] = df_encoded["sentiment"].apply(
    encode_sentiment
)

df_encoded.head()
```

```python
df_encoded["sentiment"].value_counts().sort_index()
```

```python
sorted(df_encoded["sentiment"].unique())
#df_encoded.dtypes
```

```python
print("Dataset original:", df.shape)
print("Dataset codificado:", df_encoded.shape)
```

```python
OUTPUT_FILE.parent.mkdir(parents=True, exist_ok=True)

df_encoded.to_csv(
    OUTPUT_FILE,
    index=False
)

print(f"Dataset codificado salvo em:\n{OUTPUT_FILE}")
```
