# Normalização das Avaliações do IMDb

Este notebook apresenta o processo de normalização textual aplicado ao
dataset de avaliações de filmes do IMDb.

## Objetivo

Preparar os textos para as etapas posteriores de processamento,
encoding e representação numérica.

## Etapas realizadas

- Conversão para letras minúsculas;
- Remoção de tags HTML;
- Tratamento de entidades HTML;
- Normalização dos espaços em branco;
- Remoção de espaços extras no início e no final.

```python
import sys
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT / "src") not in sys.path:
    sys.path.append(str(PROJECT_ROOT / "src"))

from preprocessing.normalization import normalize_text
```

```python
RAW_DATA = PROJECT_ROOT / "data" / "raw" / "IMDB_Dataset_menor.csv"
PROCESSED_DATA = (
    PROJECT_ROOT / "data" / "processed" / "IMDB_Dataset_menor_normalizado.csv"
)

print("Dataset original:", RAW_DATA)
print("Dataset processado:", PROCESSED_DATA)
```

```python
df = pd.read_csv(RAW_DATA)

print("Quantidade de linhas:", len(df))
print("Quantidade de colunas:", len(df.columns))

df.head()
```

```python
df.info()
df.columns
df.isnull().sum()
df["sentiment"].value_counts()
```

```python
for i in range(3):
    original = df.loc[i, "review"]
    normalizado = normalize_text(original)

    print(f"--- Avaliação {i + 1} ---")
    print("ORIGINAL:")
    print(original)

    print("\nNORMALIZADO:")
    print(normalizado)

    print("\n" + "=" * 80 + "\n")
```

```python
df_normalizado = df.copy()

df_normalizado["review"] = df_normalizado["review"].apply(normalize_text)
```

```python
df_normalizado["review"].head(10)
#print("Original:", df.shape)
#print("Normalizado:", df_normalizado.shape)
```

```python
PROCESSED_DATA.parent.mkdir(parents=True, exist_ok=True)

df_normalizado.to_csv(
    PROCESSED_DATA,
    index=False
)

print(f"Dataset normalizado salvo em:")
print(PROCESSED_DATA)
```
